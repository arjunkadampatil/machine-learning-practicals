# Practical: Simple Linear Regression

**Dataset:** Iris Dataset (Setosa class)  
**Subject:** Machine Learning

This practical demonstrates Simple Linear Regression using:
1. Statistical Formula Method
2. Matrix (Normal Equation) Method


## Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

## Load Iris Dataset

```python
iris = load_iris()

# Using first 50 samples (Setosa class)
X = iris.data[:50, 0]   # Sepal Length
Y = iris.data[:50, 1]   # Sepal Width
```

## Method 1: Simple Linear Regression (Statistical Method)

```python
x_mean = np.mean(X)
y_mean = np.mean(Y)

cov_xy = np.mean((X - x_mean) * (Y - y_mean))
var_x = np.mean((X - x_mean) ** 2)

m1 = cov_xy / var_x
c1 = y_mean - m1 * x_mean

print("Slope (m):", m1)
print("Intercept (c):", c1)
```

## Plot: Statistical Method

```python
Y_pred_1 = m1 * X + c1

plt.scatter(X, Y, label='Actual Data')
plt.plot(X, Y_pred_1, color='red', label='Regression Line')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('Simple Linear Regression (Statistical Method)')
plt.legend()
plt.grid(True)
plt.show()
```

## Method 2: Simple Linear Regression (Matrix Method)

```python
X_matrix = X.reshape(-1, 1)
Y_matrix = Y.reshape(-1, 1)

X_design = np.hstack((np.ones((X_matrix.shape[0], 1)), X_matrix))

theta = np.linalg.inv(X_design.T @ X_design) @ X_design.T @ Y_matrix

c2 = theta[0][0]
m2 = theta[1][0]

print("Slope (m):", m2)
print("Intercept (c):", c2)
```

## Plot: Matrix Method

```python
Y_pred_2 = m2 * X + c2

plt.scatter(X, Y, label='Actual Data')
plt.plot(X, Y_pred_2, color='green', label='Regression Line')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('Simple Linear Regression (Matrix Method)')
plt.legend()
plt.grid(True)
plt.show()
```

## Conclusion

Simple Linear Regression was successfully implemented using both statistical and matrix methods, and the relationship between variables was visualized using plots.
